# Probability

## 1. Introduction to Probability

### Understand the Concept

Probability is a measure of how likely an event is to happen.

The value of probability is always between 0 and 1.

- Probability = 0 → Event is impossible.
- Probability = 1 → Event is certain.
- Probability between 0 and 1 → Event has some chance of happening.

### Formula

P(A) = Number of favorable outcomes / Total number of possible outcomes

Where:

- P(A) = Probability of event A
- Favorable outcomes = Outcomes that satisfy the event
- Total outcomes = All possible outcomes

### Numerical Example

Suppose we toss a fair coin.

Possible outcomes:

Heads, Tails

There is 1 favorable outcome for getting Heads and 2 total outcomes.

P(Heads) = 1 / 2 = 0.5

So, the probability of getting Heads is 0.5 or 50%.

### Demonstrate the Concept

Probability helps us measure uncertainty and estimate the chance of different outcomes.

```python
# Probability of getting Heads when tossing a fair coin

favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes

print("Probability of getting Heads:", probability)
```

### Business Example

A company can use probability to estimate the chance of a customer making a purchase.

For example, if historical data shows that 30 out of 100 customers make a purchase, the estimated probability is:

30 / 100 = 0.3

So, the probability of a customer making a purchase is 30%.

### Interpretation

A probability of 0.3 means there is an estimated 30% chance that the event will happen.

### AI/ML Use Case

Probability is important in AI/ML because many models predict the probability of different outcomes.

For example, a classification model can predict the probability that a customer will churn or not churn.

## 2. Events

### Understand the Concept

An event is a specific outcome or a group of outcomes that we are interested in when performing a random experiment.

An event is usually represented by a capital letter such as A or B.

For example, when rolling a die, getting an even number is an event.

Event A = {2, 4, 6}

### Formula

The probability of an event is:

P(A) = Number of favorable outcomes / Total number of possible outcomes

### Numerical Example

Suppose we roll a fair six-sided die.

Sample space:

S = {1, 2, 3, 4, 5, 6}

Let A be the event of getting an even number.

A = {2, 4, 6}

There are 3 favorable outcomes and 6 total outcomes.

P(A) = 3 / 6 = 0.5

Therefore, the probability of getting an even number is 0.5 or 50%.

### Demonstrate the Concept

An event can contain one outcome or multiple outcomes depending on what we are interested in.

```python
# Probability of getting an even number when rolling a fair die

sample_space = [1, 2, 3, 4, 5, 6]
event = [2, 4, 6]

probability = len(event) / len(sample_space)

print("Probability of getting an even number:", probability)
```

### Business Example

A company can define an event as whether a customer purchases a product.

For example:

Event A = Customer makes a purchase.

The company can calculate the probability of this event using historical customer data.

### Interpretation

An event represents the outcome or group of outcomes we want to study.

The probability tells us how likely that event is to occur.

### AI/ML Use Case

Events are used in classification and prediction problems.

For example, in fraud detection:

Event A = Transaction is fraudulent.

A machine learning model can estimate the probability of this event.

## 3. Sample Space

### Understand the Concept

A sample space is the complete set of all possible outcomes of a random experiment.

It is usually represented by the letter S.

For example, when tossing a coin, the possible outcomes are Heads and Tails.

Sample Space:

S = {Heads, Tails}

### Formula

For an event A:

P(A) = Number of favorable outcomes / Total number of outcomes in the sample space

### Numerical Example

Suppose we roll a fair six-sided die.

The sample space is:

S = {1, 2, 3, 4, 5, 6}

There are 6 possible outcomes.

If event A is getting a number greater than 4:

A = {5, 6}

Therefore:

P(A) = 2 / 6 = 1 / 3 ≈ 0.333

### Demonstrate the Concept

The sample space gives us all possible outcomes before calculating the probability of a particular event.

```python
# Sample space for rolling a fair die

sample_space = [1, 2, 3, 4, 5, 6]

event = [value for value in sample_space if value > 4]

probability = len(event) / len(sample_space)

print("Sample Space:", sample_space)
print("Favorable Outcomes:", event)
print("Probability:", probability)
```

### Business Example

A company can define a sample space when analyzing possible customer outcomes.

For example, a customer interaction may have outcomes such as:

{Purchase, No Purchase}

These outcomes together form the sample space.

### Interpretation

The sample space contains every possible outcome of the experiment.

Knowing the complete sample space helps us correctly calculate probabilities.

### AI/ML Use Case

Sample spaces are useful when defining possible outcomes in classification and probabilistic models.

For example, a classification problem may have the sample space:

{Fraud, Not Fraud}

## 4. Independent Events

### Understand the Concept

Two events are independent when the occurrence of one event does not affect the probability of the other event.

For example, tossing a coin twice. The result of the first toss does not affect the result of the second toss.

### Formula

For independent events A and B:

P(A ∩ B) = P(A) × P(B)

Where:

- P(A ∩ B) = Probability that both A and B occur
- P(A) = Probability of event A
- P(B) = Probability of event B

### Numerical Example

Suppose we toss a fair coin twice.

Probability of getting Heads on the first toss:

P(A) = 1/2

Probability of getting Heads on the second toss:

P(B) = 1/2

Therefore:

P(A ∩ B) = 1/2 × 1/2 = 1/4

So, the probability of getting Heads twice is 0.25 or 25%.

### Demonstrate the Concept

The two coin tosses are independent because the result of the first toss does not change the probability of the second toss.

```python
# Probability of getting Heads twice in two independent coin tosses

p_first_heads = 1 / 2
p_second_heads = 1 / 2

probability = p_first_heads * p_second_heads

print("Probability of getting Heads twice:", probability)
```

### Business Example

A company may treat two unrelated customer events as independent if one event does not affect the other.

For example, two separate customers making purchases may be considered independent events when one customer's purchase does not affect the other customer's decision.

### Interpretation

For independent events, the probability of both events happening is found by multiplying their individual probabilities.

### AI/ML Use Case

Independent events are important in probabilistic models and algorithms that make assumptions about relationships between variables.

For example, the Naive Bayes algorithm assumes that features are conditionally independent given the class.

## 5. Dependent Events

### Understand the Concept

Two events are dependent when the occurrence of one event affects the probability of the other event.

A common example is drawing cards from a deck without replacing the first card.

### Formula

For dependent events A and B:

P(A ∩ B) = P(A) × P(B | A)

Where:

- P(A ∩ B) = Probability that both A and B occur
- P(A) = Probability of event A
- P(B | A) = Probability of B occurring when A has already occurred

### Numerical Example

Suppose a bag contains 3 red balls and 2 blue balls.

There are 5 balls in total.

If we draw one red ball and do not replace it:

P(A) = 3 / 5

Now, 2 red balls remain out of 4 balls.

P(B | A) = 2 / 4

Therefore:

P(A ∩ B) = 3/5 × 2/4
          = 6/20
          = 0.3

So, the probability of drawing two red balls without replacement is 0.3 or 30%.

### Demonstrate the Concept

The events are dependent because after the first red ball is removed, the number of remaining red balls and total balls changes.

```python
# Probability of drawing two red balls without replacement

p_first_red = 3 / 5
p_second_red = 2 / 4

probability = p_first_red * p_second_red

print("Probability of drawing two red balls:", probability)
```

### Business Example

A company can have dependent events when one business event changes the probability of another event.

For example, if a limited number of products are available, selling one product reduces the number of products available for the next customer.

### Interpretation

In dependent events, the first event changes the conditions for the second event.

Therefore, the probability of the second event depends on what happened in the first event.

### AI/ML Use Case

Dependent relationships are important when analyzing real-world data because features and events may influence each other.

Understanding dependencies helps in building better probabilistic models and analyzing relationships between variables.

## 6. Conditional Probability

### Understand the Concept

Conditional probability is the probability of an event happening when we already know that another event has happened.

It is useful when the occurrence of one event gives us additional information about another event.

### Formula

The conditional probability of A given B is:

P(A | B) = P(A ∩ B) / P(B)

Where:

- P(A | B) = Probability of A given B
- P(A ∩ B) = Probability that both A and B occur
- P(B) = Probability of event B

### Numerical Example

Suppose a class has 20 students.

- 12 students know Python.
- 8 students know both Python and SQL.

Let:

A = Student knows SQL

B = Student knows Python

Then:

P(A | B) = P(A ∩ B) / P(B)

P(A | B) = (8/20) / (12/20)

P(A | B) = 8/12

P(A | B) ≈ 0.667

So, the probability that a student knows SQL given that the student knows Python is approximately 66.7%.

### Demonstrate the Concept

Conditional probability updates the probability of an event based on information that another event has already occurred.

```python
# Conditional probability example

students_who_know_python = 12
students_who_know_both = 8

probability = students_who_know_both / students_who_know_python

print(
    "Probability of knowing SQL given Python:",
    probability
)
```

### Business Example

A company can calculate the probability that a customer will purchase a product given that the customer has clicked on an advertisement.

For example:

P(Purchase | Ad Click)

This tells the company how likely a customer is to purchase after clicking the advertisement.

### Interpretation

Conditional probability allows us to calculate probability using additional information.

The probability of an event can change when we know that another event has already occurred.

### AI/ML Use Case

Conditional probability is widely used in classification and probabilistic machine learning.

For example, a model can estimate:

P(Spam | Email Features)

This represents the probability that an email is spam given the observed features.

## 7. Bayes' Theorem

### Understand the Concept

Bayes' Theorem is used to calculate the probability of an event based on new or additional information.

It helps us update an existing probability when new evidence is available.

### Formula

P(A | B) = [P(B | A) × P(A)] / P(B)

Where:

- P(A | B) = Probability of A given B
- P(B | A) = Probability of B given A
- P(A) = Prior probability of A
- P(B) = Probability of the evidence B

### Numerical Example

Suppose:

- 1% of customers are fraudulent.
- A fraud detection system identifies a fraudulent transaction 90% of the time.
- The system gives a positive result for 5% of all transactions.

Let:

A = Transaction is fraudulent

B = System gives a positive result

Then:

P(A) = 0.01
P(B | A) = 0.90
P(B) = 0.05

Using Bayes' Theorem:

P(A | B) = (0.90 × 0.01) / 0.05

P(A | B) = 0.18

So, the probability that a transaction is actually fraudulent given a positive result is 18%.

### Demonstrate the Concept

Bayes' Theorem combines prior probability with new evidence to calculate an updated probability.

```python
# Bayes' Theorem example

p_fraud = 0.01
p_positive_given_fraud = 0.90
p_positive = 0.05

p_fraud_given_positive = (
    p_positive_given_fraud * p_fraud
) / p_positive

print("Probability of fraud given a positive result:",
      p_fraud_given_positive)
```

### Business Example

Banks and financial companies can use Bayes' Theorem to update the probability of a fraudulent transaction when new transaction information becomes available.

For example, the probability of fraud can be updated using information such as transaction amount, location, or unusual activity.

### Interpretation

Bayes' Theorem helps us update a probability after receiving new evidence.

In the example, even though only 1% of transactions are fraudulent, a positive detection result increases the probability of fraud to 18%.

### AI/ML Use Case

Bayes' Theorem is used in probabilistic machine learning.

For example, Naive Bayes classification uses Bayes' Theorem to calculate the probability of different classes based on observed features.

## 8. Probability Rules

### Understand the Concept

Probability rules help us calculate the probability of events and combinations of events.

Some important probability rules are:

1. Complement Rule

P(Aᶜ) = 1 - P(A)

2. Addition Rule

For mutually exclusive events:

P(A ∪ B) = P(A) + P(B)

3. Multiplication Rule

For independent events:

P(A ∩ B) = P(A) × P(B)

### Numerical Example

Suppose:

P(A) = 0.6

Then, using the complement rule:

P(Aᶜ) = 1 - 0.6
       = 0.4

So, the probability that event A does not occur is 0.4 or 40%.

For two mutually exclusive events:

P(A) = 0.3
P(B) = 0.2

P(A ∪ B) = 0.3 + 0.2
          = 0.5

### Demonstrate the Concept

These rules help us calculate probabilities when dealing with single events, combined events, or events that do not occur.

```python
# Probability Rules

p_a = 0.6

# Complement Rule
p_not_a = 1 - p_a

print("P(Not A):", p_not_a)

# Addition Rule for mutually exclusive events
p_a = 0.3
p_b = 0.2

p_a_or_b = p_a + p_b

print("P(A or B):", p_a_or_b)

# Multiplication Rule for independent events
p_a = 0.5
p_b = 0.4

p_a_and_b = p_a * p_b

print("P(A and B):", p_a_and_b)
```

### Business Example

A company can use probability rules to analyze customer behavior.

For example, if the probability of a customer purchasing a product is 0.6, the probability of the customer not purchasing it is:

1 - 0.6 = 0.4

### Interpretation

Probability rules provide different ways to calculate the chance of events occurring, not occurring, or occurring together.

### AI/ML Use Case

Probability rules are used in probabilistic models, classification, risk analysis, and decision-making systems.

They help models combine probabilities and estimate the likelihood of different outcomes.

## 9. Random Variables

### Understand the Concept

A random variable is a variable that assigns a numerical value to the outcome of a random experiment.

There are two main types:

- Discrete Random Variable – takes countable values.
- Continuous Random Variable – can take any value within a range.

For example, the number of customers visiting a store is a discrete random variable, while a customer's waiting time can be a continuous random variable.

### Formula

For a discrete random variable, the probability distribution satisfies:

Σ P(X = x) = 1

This means the probabilities of all possible values must add up to 1.

### Numerical Example

Suppose X represents the number of heads when tossing two coins.

Possible values:

X = 0, 1, 2

The probabilities are:

P(X = 0) = 1/4

P(X = 1) = 2/4

P(X = 2) = 1/4

Total probability:

1/4 + 2/4 + 1/4 = 1

### Demonstrate the Concept

A random variable converts random outcomes into numerical values so that we can analyze them mathematically.

```python
# Probability distribution of the number of heads
# when tossing two fair coins

values = [0, 1, 2]
probabilities = [1/4, 2/4, 1/4]

for value, probability in zip(values, probabilities):
    print(f"P(X = {value}) = {probability}")

print("Total Probability:", sum(probabilities))
```

### Business Example

A company can use a random variable to represent the number of products sold in a day.

For example:

X = Number of products sold in one day

The value of X changes depending on the day's sales.

### Interpretation

A random variable represents the numerical outcome of a random experiment.

Its probability distribution tells us how likely each possible value is.

### AI/ML Use Case

Random variables are fundamental in probability and machine learning.

They are used to represent uncertain quantities such as customer demand, transaction amounts, prediction outcomes, and other data variables.

## 10. Expected Value

### Understand the Concept

Expected value is the average outcome we expect from a random experiment if it is repeated many times.

It is calculated by multiplying each possible outcome by its probability and adding all the results.

### Formula

For a discrete random variable:

E(X) = Σ [x × P(X = x)]

Where:

- E(X) = Expected value
- x = Possible outcome
- P(X = x) = Probability of the outcome

### Numerical Example

Suppose a game has the following outcomes:

- Win ₹100 with probability 0.2
- Win ₹50 with probability 0.3
- Win ₹0 with probability 0.5

Expected Value:

E(X) = (100 × 0.2) + (50 × 0.3) + (0 × 0.5)

E(X) = 20 + 15 + 0

E(X) = ₹35

So, the expected value is ₹35.

### Demonstrate the Concept

Expected value does not mean that we will actually receive ₹35 in one game.

It represents the average expected outcome over many repeated trials.

```python
# Expected Value example

outcomes = [100, 50, 0]
probabilities = [0.2, 0.3, 0.5]

expected_value = sum(
    outcome * probability
    for outcome, probability in zip(outcomes, probabilities)
)

print("Expected Value: ₹", expected_value)
```

### Business Example

A company can use expected value to estimate the average revenue from different customer outcomes.

For example, if different sales outcomes have different probabilities, the company can calculate the expected revenue before making a business decision.

### Interpretation

The expected value represents the probability-weighted average of all possible outcomes.

In our example, the expected value is ₹35.

### AI/ML Use Case

Expected value is used in machine learning and decision-making to estimate the average outcome of uncertain events.

For example, it can be used in risk analysis, recommendation systems, and probabilistic predictions.
