## 1. Mean

### Understand the Concept

Mean is the average value of a group of numbers.

It is calculated by adding all the values and dividing the total by the number of values.

### Formula

Mean = Sum of all values / Number of values

### Demonstrate the Concept

Suppose the marks of five students are:

60, 70, 80, 90, 100

Mean = (60 + 70 + 80 + 90 + 100) / 5

Mean = 80

So, the average mark is 80.

```python
import numpy as np

marks = [60, 70, 80, 90, 100]

mean_marks = np.mean(marks)

print("Mean marks:", mean_marks)
```

### Interpretation

The mean of the marks is 80.

This means the average mark of the five students is 80.

### AI/ML Use Case

Mean is commonly used in data analysis to understand the average value of a feature.

For example, we can calculate the average salary, average age, or average sales in a dataset.

Mean can also be used during data preprocessing, such as replacing missing numerical values with the mean.

### Business Example

A company can calculate the mean monthly sales to understand its average sales performance.

For example, if monthly sales are ₹10,000, ₹12,000, ₹14,000, ₹16,000, and ₹18,000, the mean sales value helps the company understand its average monthly sales.

## 2. Median

### Understand the Concept

Median is the middle value of a dataset when the values are arranged in ascending or descending order.

If there is an odd number of values, the middle value is the median.

If there is an even number of values, the median is the average of the two middle values.

### Demonstrate the Concept

Suppose the marks are:

60, 70, 80, 90, 100

The values are already arranged in ascending order.

The middle value is 80.

Therefore, the median is 80.

### Even Number Example

For the values:

60, 70, 80, 90

The two middle values are 70 and 80.

Median = (70 + 80) / 2 = 75

```python
import numpy as np

marks = [60, 70, 80, 90, 100]

median_marks = np.median(marks)

print("Median marks:", median_marks)
```

### Interpretation

The median of the marks is 80 because 80 is the middle value after arranging the data.

Median is useful when the dataset contains extreme values or outliers because it is less affected by them compared to the mean.

### AI/ML Use Case

Median can be used to understand the center of numerical features.

It is also commonly used to handle missing values in a dataset, especially when the feature contains outliers.

### Business Example

A company can use the median salary to understand the typical salary of its employees.

Median is useful when a few employees have very high salaries because those extreme values can affect the mean.

## 3. Mode

### Understand the Concept

Mode is the value that appears most frequently in a dataset.

A dataset can have one mode, more than one mode, or no mode if all values occur only once.

### Demonstrate the Concept

Suppose the marks are:

60, 70, 80, 80, 90, 80, 100

The value 80 appears three times, which is more frequent than the other values.

Therefore, the mode is 80.

```python
import numpy as np

marks = [60, 70, 80, 80, 90, 80, 100]

values, counts = np.unique(marks, return_counts=True)

mode = values[np.argmax(counts)]

print("Mode:", mode)
```

### Interpretation

The mode is 80 because it occurs three times in the dataset.

Mode is useful for finding the most common value or category in data.

### AI/ML Use Case

Mode can be used to understand the most common category in a dataset.

For example, it can help identify the most common product purchased, most common customer category, or most frequent value in categorical data.

### Business Example

A retail store can use mode to identify the most frequently purchased product.

For example, if a store sells more T-shirts than shirts, shoes, and bags, T-shirt is the mode because it is the most frequently purchased product.

## 4. Mean vs Median vs Mode

### Understand the Concept

Mean, median, and mode are three common measures of central tendency.

- **Mean** → Average value of the data.
- **Median** → Middle value after arranging the data.
- **Mode** → Most frequently occurring value.

Each measure gives us a different way to understand the center or common value of a dataset.

### Demonstrate the Concept

Consider the data:

10, 20, 20, 30, 100

- Mean = (10 + 20 + 20 + 30 + 100) / 5 = 36
- Median = 20
- Mode = 20

Here, the mean is affected by the high value 100, while the median and mode remain 20.

```python
import numpy as np

data = [10, 20, 20, 30, 100]

mean = np.mean(data)
median = np.median(data)

values, counts = np.unique(data, return_counts=True)
mode = values[np.argmax(counts)]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

### Interpretation

The mean is 36, the median is 20, and the mode is 20.

The value 100 is an extreme value, so it increases the mean.

This shows that the median can be more useful than the mean when the data contains outliers.

### AI/ML Use Case

Mean, median, and mode are useful during Exploratory Data Analysis (EDA).

They help us understand the central value of features and choose suitable methods for data preprocessing and handling missing values.

## 5. Weighted Mean

### Understand the Concept

Weighted mean is an average where different values are given different levels of importance or weight.

In a normal mean, every value has the same importance.

In a weighted mean, some values contribute more to the final result than others.

### Formula

Weighted Mean = Σ(Value × Weight) / Σ(Weights)

### Demonstrate the Concept

Suppose a student's final grade is calculated using:

- Assignment = 80, Weight = 20%
- Exam = 90, Weight = 80%

Weighted Mean:

(80 × 0.20) + (90 × 0.80) = 88

So, the weighted average is 88.

```python
scores = [80, 90]
weights = [0.20, 0.80]

weighted_mean = sum(score * weight for score, weight in zip(scores, weights))

print("Weighted mean:", weighted_mean)
```

### Interpretation

The weighted mean is 88.

The exam has a higher weight, so the exam score has a greater effect on the final result.

### AI/ML Use Case

Weighted mean can be useful when different observations or features have different importance.

For example, in business analytics, recent sales data may be given more importance than older sales data when calculating an average.

### Business Example

A company can use weighted mean to calculate an overall performance score when different factors have different importance.

For example, employee performance may be calculated using 40% project performance, 30% attendance, and 30% customer feedback.

## 6. Geometric Mean

### Understand the Concept

Geometric mean is a type of average that is mainly used for values that are multiplied together or represent growth rates.

It is calculated by multiplying all the values and taking the nth root, where n is the number of values.

### Formula

Geometric Mean = ⁿ√(x₁ × x₂ × ... × xₙ)

### Demonstrate the Concept

Suppose the values are:

2, 8

Geometric Mean = √(2 × 8)

Geometric Mean = √16

Geometric Mean = 4

```python
import numpy as np

values = [2, 8]

geometric_mean = np.prod(values) ** (1 / len(values))

print("Geometric mean:", geometric_mean)
```

### Interpretation

The geometric mean of 2 and 8 is 4.

It is different from the arithmetic mean because it uses multiplication instead of addition.

### AI/ML Use Case

Geometric mean can be useful when analyzing growth rates, ratios, or values that change multiplicatively.

For example, it can be used to summarize investment growth rates or performance changes over multiple periods.

### Business Example

A business can use geometric mean to calculate the average growth rate over multiple years.

For example, if a company's sales grow by different percentages each year, geometric mean can be used to find the overall average growth rate.

## 7. Harmonic Mean

### Understand the Concept

Harmonic mean is a type of average that is useful for rates and ratios.

It is calculated by dividing the number of values by the sum of their reciprocals.

### Formula

Harmonic Mean = n / (1/x₁ + 1/x₂ + ... + 1/xₙ)

### Demonstrate the Concept

Suppose a car travels at:

- 60 km/h for one part of a journey
- 40 km/h for another equal-distance part

The harmonic mean is:

2 / (1/60 + 1/40) = 48 km/h

So, the average speed is 48 km/h for equal distances.

```python
import numpy as np

speeds = [60, 40]

harmonic_mean = len(speeds) / sum(1 / speed for speed in speeds)

print("Harmonic mean:", harmonic_mean)
```

### Interpretation

The harmonic mean of 60 and 40 is 48 km/h.

Harmonic mean is useful when working with rates, ratios, or quantities such as speed.

### AI/ML Use Case

Harmonic mean is used in some machine learning evaluation metrics.

For example, the F1-score is the harmonic mean of precision and recall. It gives a balanced measure when both precision and recall are important.

### Business Example

A business can use harmonic mean when calculating average rates.

For example, if a delivery company measures delivery speed across different routes, harmonic mean can be useful for calculating an average rate when the distances are equal.
