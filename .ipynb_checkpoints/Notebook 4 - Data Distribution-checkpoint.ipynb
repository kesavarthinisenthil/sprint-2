## 1. Normal Distribution

### Understand the Concept

A normal distribution is a continuous probability distribution that is symmetric and bell-shaped.

In a normal distribution, most values are concentrated around the mean, and fewer values occur as we move away from the mean.

For a normal distribution:

Mean = Median = Mode

### Formula

The probability density function is:

f(x) = (1 / (σ√(2π))) × e^(-(x - μ)² / (2σ²))

Where:

- μ = Mean
- σ = Standard deviation
- x = Data value

### Demonstrate the Concept

Suppose the marks of students approximately follow a normal distribution with:

Mean = 70

Standard deviation = 10

Most students will have marks close to 70, while fewer students will have very low or very high marks.

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 70
std = 10

x = np.linspace(mean - 4 * std, mean + 4 * std, 200)

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -((x - mean) ** 2) / (2 * std ** 2)
)

plt.plot(x, y)
plt.xlabel("Marks")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

### Business Example

A company can use a normal distribution to analyze employee performance scores when the scores are approximately normally distributed.

For example, most employee scores may be close to the average, with fewer employees having very low or very high scores.

### Interpretation

The graph has a bell-shaped and symmetric pattern.

Most values are concentrated around the mean, and the frequency decreases as we move away from the mean.

### AI/ML Use Case

Normal distribution is commonly used in statistics and machine learning.

It helps us understand feature distributions and is used in methods such as statistical modeling, hypothesis testing, and standardization.

## 2. Uniform Distribution

### Understand the Concept

A uniform distribution is a probability distribution where all values within a given range have an equal chance of occurring.

It can be:
- Continuous — any value within the range can occur.
- Discrete — specific values have equal probability.

### Formula

For a continuous uniform distribution:

f(x) = 1 / (b - a)

Where:

- a = Lower limit
- b = Upper limit
- x = Value between a and b

### Numerical Example

Suppose values are uniformly distributed between 10 and 20.

Here:

a = 10
b = 20

Therefore:

f(x) = 1 / (20 - 10)
     = 1 / 10
     = 0.1

This means the probability density is constant throughout the range from 10 to 20.

### Demonstrate the Concept

In a uniform distribution, the graph is approximately flat because values in the given range have equal probability.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(10, 20, 1000)

plt.hist(data, bins=20)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

### Business Example

A company can use a uniform distribution in simulations when values within a fixed range are assumed to have equal chances.

For example, if a random processing time is assumed to be between 10 and 20 minutes, each value in that range can be simulated with equal probability.

### Interpretation

The histogram is approximately flat across the range.

This shows that values between 10 and 20 have roughly equal chances of occurring.

### AI/ML Use Case

Uniform distribution can be used in simulations, random data generation, and testing algorithms with values from a fixed range.

## 3. Gaussian Distribution

### Understand the Concept

Gaussian distribution is another name commonly used for the normal distribution.

It is a continuous probability distribution with a bell-shaped and symmetric curve.

In a Gaussian distribution:

- Most values are close to the mean.
- Fewer values occur farther from the mean.
- Mean = Median = Mode.

### Formula

The probability density function is:

f(x) = (1 / (σ√(2π))) × e^(-(x - μ)² / (2σ²))

Where:

- μ = Mean
- σ = Standard deviation
- x = Data value

### Numerical Example

Suppose:

Mean (μ) = 50  
Standard deviation (σ) = 5

For x = 50:

f(50) = 1 / (5√(2π))

≈ 0.0798

The highest density occurs around the mean.

### Demonstrate the Concept

A Gaussian distribution can be represented using a bell-shaped curve centered around its mean.

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std = 5

x = np.linspace(mean - 4 * std, mean + 4 * std, 200)

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -((x - mean) ** 2) / (2 * std ** 2)
)

plt.plot(x, y)
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Gaussian Distribution")
plt.show()
```

### Business Example

A company can use a Gaussian distribution to understand employee performance scores when the scores are approximately normally distributed.

For example, most employee scores may be close to the average score, while very low and very high scores occur less often.

### Interpretation

The graph is bell-shaped and symmetric around the mean.

The density is highest near the mean and decreases as we move away from it.

### AI/ML Use Case

Gaussian distribution is used in statistical modeling, feature analysis, hypothesis testing, and algorithms that assume or work well with normally distributed data.

## 4. Left Skew

### Understand the Concept

Left skew is also called negative skewness.

In a left-skewed distribution, the long tail extends toward the left side.

Most of the values are concentrated on the right side, while a few smaller values create the long left tail.

Usually:

Mean < Median < Mode

### Numerical Example

Consider the data:

2, 8, 9, 10, 10, 11, 12, 13, 14

Here, the value 2 is much smaller than the other values and pulls the mean toward the left.

Mean ≈ 9.89

Median = 10

This shows the mean is lower than the median, which is a common sign of left skew.

### Demonstrate the Concept

A left-skewed distribution has most values on the right and a longer tail extending toward the left.

```python
import numpy as np
import matplotlib.pyplot as plt

data = [2, 8, 9, 10, 10, 11, 12, 13, 14]

plt.hist(data, bins=6)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Left-Skewed Distribution")
plt.show()
```

### Business Example

A company may analyze employee retirement ages. If most employees retire at older ages but a few retire much earlier, the distribution can be left-skewed.

### Interpretation

The distribution has a longer tail toward the left.

A few smaller values pull the mean toward the lower side.

### AI/ML Use Case

Skewness helps identify the shape of feature distributions during data analysis.

If a feature is strongly left-skewed, transformation or other preprocessing methods may be considered before using it in some machine learning models.

## 5. Right Skew

### Understand the Concept

Right skew is also called positive skewness.

In a right-skewed distribution, the long tail extends toward the right side.

Most of the values are concentrated on the left side, while a few larger values create the long right tail.

Usually:

Mean > Median > Mode

### Numerical Example

Consider the data:

1, 2, 3, 4, 5, 5, 6, 7, 20

Here, the value 20 is much larger than the other values and pulls the mean toward the right.

Mean ≈ 5.89

Median = 5

This shows the mean is higher than the median, which is a common sign of right skew.

### Demonstrate the Concept

A right-skewed distribution has most values on the left and a longer tail extending toward the right.

```python
import numpy as np
import matplotlib.pyplot as plt

data = [1, 2, 3, 4, 5, 5, 6, 7, 20]

plt.hist(data, bins=6)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Right-Skewed Distribution")
plt.show()
```

### Business Example

A company may analyze employee salaries. Most employees may have salaries within a lower or middle range, while a small number of employees have very high salaries.

This can create a right-skewed salary distribution.

### Interpretation

The distribution has a longer tail toward the right.

A few large values pull the mean toward the higher side.

### AI/ML Use Case

Right-skewed features are common in real-world datasets, such as income, sales, and transaction amounts.

Identifying skewness helps during data preprocessing and can help decide whether a transformation is useful.

## 6. Symmetric Distribution

### Understand the Concept

A symmetric distribution is a distribution where the left and right sides have a similar shape around the center.

The values are distributed evenly around the central point.

In a perfectly symmetric distribution:

Mean = Median = Mode

### Numerical Example

Consider the data:

2, 4, 6, 8, 10, 12, 14

Here:

Mean = 8

Median = 8

Mode = No mode

The values are balanced around 8, so the distribution is symmetric.

### Demonstrate the Concept

A symmetric distribution has similar patterns on both sides of the center.

The normal distribution is a common example of a symmetric distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=20)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Symmetric Distribution")
plt.show()
```

### Business Example

A company can analyze employee performance scores when most scores are distributed evenly around the average.

For example, if employees' scores are balanced around the average score, the distribution may be approximately symmetric.

### Interpretation

The distribution has a similar shape on both sides of the center.

There is no long tail toward either side.

### AI/ML Use Case

Understanding whether a feature is symmetric helps during exploratory data analysis and preprocessing.

Symmetric features can be easier to analyze using common statistical methods, especially when the distribution is approximately normal.

## 7. Kurtosis

### Understand the Concept

Kurtosis is a statistical measure that describes the heaviness of the tails of a distribution compared with a normal distribution.

It helps us understand how likely extreme values are in the data.

There are three common types:

- Mesokurtic – similar tail behavior to a normal distribution.
- Leptokurtic – heavier tails and more extreme values.
- Platykurtic – lighter tails and fewer extreme values.

### Formula

Kurtosis can be calculated as:

K = E[(X - μ)⁴] / σ⁴

Where:

- X = Data value
- μ = Mean
- σ = Standard deviation

For a normal distribution:

K = 3

Excess kurtosis is:

Excess Kurtosis = K - 3

Therefore, a normal distribution has excess kurtosis = 0.

### Numerical Example

Suppose a dataset has:

Kurtosis = 5

Since 5 > 3, the distribution has heavier tails than a normal distribution.

Its excess kurtosis is:

5 - 3 = 2

This indicates a higher possibility of extreme values compared with a normal distribution.

### Demonstrate the Concept

Kurtosis mainly helps us understand the tails and extreme values of a distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

mean = np.mean(data)
std = np.std(data)

kurtosis = np.mean((data - mean) ** 4) / (std ** 4)
excess_kurtosis = kurtosis - 3

print("Kurtosis:", kurtosis)
print("Excess Kurtosis:", excess_kurtosis)

plt.hist(data, bins=20)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Kurtosis of a Distribution")
plt.show()
```

### Business Example

A financial company can use kurtosis to study investment returns.

High kurtosis can indicate heavier tails, meaning extreme gains or losses may occur more often than in a normal distribution.

### Interpretation

Kurtosis helps us understand the tail behavior of the data.

- Positive excess kurtosis → heavier tails and more extreme values.
- Zero excess kurtosis → similar tail behavior to a normal distribution.
- Negative excess kurtosis → lighter tails and fewer extreme values.

### AI/ML Use Case

Kurtosis can be used during exploratory data analysis to identify features with heavy tails and possible extreme values.

It can help in understanding data distribution before applying preprocessing or machine learning models.

## 8. Empirical Rule (68–95–99.7)

### Understand the Concept

The Empirical Rule describes how data is distributed around the mean in an approximately normal distribution.

It states that:

- About 68% of the data falls within 1 standard deviation of the mean.
- About 95% of the data falls within 2 standard deviations of the mean.
- About 99.7% of the data falls within 3 standard deviations of the mean.

### Formula

The ranges are:

68% → μ ± 1σ

95% → μ ± 2σ

99.7% → μ ± 3σ

Where:

- μ = Mean
- σ = Standard deviation

### Numerical Example

Suppose:

Mean = 70

Standard deviation = 10

For 1 standard deviation:

70 ± 10 → 60 to 80

Approximately 68% of values fall between 60 and 80.

For 2 standard deviations:

70 ± 20 → 50 to 90

Approximately 95% of values fall between 50 and 90.

For 3 standard deviations:

70 ± 30 → 40 to 100

Approximately 99.7% of values fall between 40 and 100.

### Demonstrate the Concept

The Empirical Rule shows that most observations in an approximately normal distribution are close to the mean, and very few observations are far from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 70
std = 10

x = np.linspace(mean - 4 * std, mean + 4 * std, 1000)

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -((x - mean) ** 2) / (2 * std ** 2)
)

plt.plot(x, y)
plt.axvline(mean - std)
plt.axvline(mean + std)
plt.axvline(mean - 2 * std)
plt.axvline(mean + 2 * std)
plt.axvline(mean - 3 * std)
plt.axvline(mean + 3 * std)

plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Empirical Rule: 68–95–99.7")
plt.show()
```

### Business Example

A company can use the Empirical Rule to analyze employee performance scores when the scores are approximately normally distributed.

For example, if the average score is 70 and the standard deviation is 10, most employees' scores are expected to fall between 40 and 100.

### Interpretation

The Empirical Rule shows that:

- 68% of values are approximately within 1 standard deviation.
- 95% of values are approximately within 2 standard deviations.
- 99.7% of values are approximately within 3 standard deviations.

This helps identify values that are unusually far from the mean.

### AI/ML Use Case

The Empirical Rule can be used to understand normally distributed features and identify unusual observations.

For example, values beyond 3 standard deviations from the mean may be investigated as potential anomalies.
