## T-Test

### Definition
A T-test is a statistical test used to compare the means of one or two groups and determine whether the difference is statistically significant.

It is commonly used when the sample size is small and the population standard deviation is unknown.

### Assumptions
- Data is numerical.
- Observations are independent.
- Data is approximately normally distributed.
- For an independent two-sample T-test, the groups should be independent.
- For the standard Student's T-test, equal variance is assumed between groups; Welch's T-test can be used when variances are not equal.

### When to Use
- Comparing the mean of one sample with a known value.
- Comparing the means of two independent groups.
- Comparing measurements from the same subjects before and after an intervention.

### When Not to Use
- Data is strongly non-normal for a small sample.
- Data is categorical.
- Observations are not independent when using an independent T-test.
- A suitable non-parametric test may be preferred when assumptions are not satisfied.

### Business Example
A company wants to know whether the average sales before and after a new marketing strategy are significantly different.

### Python Example
We can use `scipy.stats.ttest_1samp()` for a one-sample T-test.

### Interpretation
If the p-value is less than the chosen significance level, we reject the Null Hypothesis and conclude that there is evidence of a statistically significant difference.

```python
from scipy import stats

sales = [52, 49, 51, 53, 50, 54, 48, 52, 51, 50]

# Test whether the average sales is different from 50
t_statistic, p_value = stats.ttest_1samp(sales, 50)

print("T-statistic:", t_statistic)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a statistically significant difference.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence of a statistically significant difference.")
```

### Business Example
A company can use a T-test to check whether the average sales after a new marketing campaign is significantly different from the previous average.

### Interpretation
- p-value < 0.05 → Reject H₀.
- p-value ≥ 0.05 → Fail to reject H₀.

The result tells us whether there is enough statistical evidence of a difference. It does not by itself tell us whether the difference is practically important.

### AI/ML Use Case
T-tests can be used to compare model performance or investigate whether a feature is associated with a statistically significant difference between groups.

## Z-Test

### Definition
A Z-test is a statistical test used to determine whether a sample mean or proportion is significantly different from a population value.

It is commonly used when the sample size is large or when the population standard deviation is known.

### Assumptions
- Data is numerical for a mean Z-test.
- Observations are independent.
- Population standard deviation is known for the standard one-sample mean Z-test.
- The sampling distribution is approximately normal.

### When to Use
- Comparing a sample mean with a known population mean.
- Comparing a sample proportion with a known population proportion.
- Large-sample hypothesis testing.

### When Not to Use
- Population standard deviation is unknown and the sample is small.
- Data does not satisfy the assumptions of the test.
- Observations are not independent.

### Formula

Z = (x̄ - μ) / (σ / √n)

Where:
- x̄ = Sample Mean
- μ = Population Mean
- σ = Population Standard Deviation
- n = Sample Size

### Business Example
A company wants to check whether the average delivery time differs from the known population average.

### Interpretation
A large absolute Z-score indicates that the sample result is far from the hypothesized population value.

```python
import math
from scipy.stats import norm

sample_mean = 52
population_mean = 50
population_std = 10
sample_size = 100

z_score = (sample_mean - population_mean) / (
    population_std / math.sqrt(sample_size)
)

p_value = 2 * (1 - norm.cdf(abs(z_score)))

print("Z-score:", z_score)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example
A company knows that the average delivery time is 50 minutes with a population standard deviation of 10 minutes. It takes a sample of deliveries to check whether the average has changed.

### Interpretation
- p-value < 0.05 → Reject H₀.
- p-value ≥ 0.05 → Fail to reject H₀.

The Z-test helps determine whether the sample result is statistically different from the population value.

### AI/ML Use Case
Z-tests can be used in statistical analysis to compare sample statistics with known population values or expected values.

## Chi-Square Test

### Definition
The Chi-Square test is a statistical test used to analyze categorical data.

It checks whether there is a significant difference between observed and expected frequencies, or whether two categorical variables are associated.

### Main Uses
1. **Chi-Square Goodness-of-Fit Test**
   - Checks whether observed frequencies match expected frequencies.

2. **Chi-Square Test of Independence**
   - Checks whether two categorical variables are related.

### Assumptions
- Data should be categorical.
- Observations should be independent.
- Expected frequencies should generally be sufficiently large.
- Each observation should belong to only one category.

### When to Use
- Testing relationships between categorical variables.
- Comparing observed and expected frequencies.
- Analyzing customer categories, product choices, survey responses, etc.

### When Not to Use
- Data is continuous and has not been converted into meaningful categories.
- Observations are not independent.
- Expected frequencies are too small for the usual Chi-Square approximation.

### Formula

χ² = Σ[(O - E)² / E]

Where:
- O = Observed Frequency
- E = Expected Frequency

### Business Example
A company wants to know whether customer preference for different products is independent of customer age group.

### Interpretation
A small p-value indicates evidence against the Null Hypothesis.

```python
from scipy.stats import chi2_contingency

# Rows = Age Groups
# Columns = Product Preferences

data = [
    [30, 20, 10],
    [20, 25, 15],
    [10, 15, 25]
]

chi2, p_value, degrees_of_freedom, expected = chi2_contingency(data)

print("Chi-Square Statistic:", chi2)
print("p-value:", p_value)
print("Degrees of Freedom:", degrees_of_freedom)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a significant association between the variables.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence of an association.")
```

### Business Example
A company can use a Chi-Square test to check whether product preference is related to customer age group.

### Interpretation
- p-value < 0.05 → Reject H₀.
- p-value ≥ 0.05 → Fail to reject H₀.

For a test of independence, rejecting H₀ provides evidence that the two categorical variables are associated.

### AI/ML Use Case
Chi-Square tests can be used during feature selection to examine the relationship between categorical features and a categorical target variable.

## 4. ANOVA

### Definition
ANOVA stands for Analysis of Variance. It is used to compare the means of **three or more groups** and check whether there is a statistically significant difference between them.

### Assumptions
- The observations should be independent.
- The dependent variable should be numerical.
- The data in each group should be approximately normally distributed.
- The groups should have similar variances.

### When to Use
- When comparing the means of 3 or more independent groups.
- Example: comparing average sales from 3 different regions.

### When Not to Use
- When there are only 2 groups, a t-test is usually used.
- When the observations are dependent or repeated.
- When the assumptions are strongly violated.

### Business Example
A company wants to compare the average monthly sales of three different store regions.

### Python Example
We can use `scipy.stats.f_oneway()` to perform one-way ANOVA.

### Interpretation
- If p-value < 0.05 → Reject the Null Hypothesis.
- If p-value >= 0.05 → Fail to Reject the Null Hypothesis.
- A significant ANOVA result means **at least one group mean is different**.
- ANOVA does not directly tell us which groups are different. A post-hoc test can be used for that.

```python
from scipy.stats import f_oneway

group_a = [20, 22, 19, 24, 21]
group_b = [30, 28, 32, 31, 29]
group_c = [25, 27, 24, 26, 28]

f_statistic, p_value = f_oneway(group_a, group_b, group_c)

print("F-statistic:", f_statistic)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("At least one group mean is significantly different.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence of a difference between group means.")
```

## Business Example and Interpretation

### Business Example
A company compares the average sales of three store regions: Region A, Region B, and Region C.

ANOVA helps determine whether the difference in their average sales is statistically significant.

### Interpretation
If the p-value is less than 0.05, we reject the Null Hypothesis and conclude that at least one region has a significantly different average sales value.

If the p-value is greater than or equal to 0.05, we do not have enough evidence to say that the group means are different.

### AI/ML Use Case
ANOVA can be used to compare the performance of different models or compare a numerical target across different categorical groups.

## 5. Mann-Whitney U Test

### Definition
The Mann-Whitney U Test is a **non-parametric statistical test** used to compare two independent groups.

It is useful when the data does not satisfy the assumptions required for an independent t-test.

### Assumptions
- The two groups should be independent.
- The data should be ordinal or numerical.
- Observations should be independent.
- If we interpret the result as a difference in medians, the groups should have a similar distribution shape.

### When to Use
- When comparing two independent groups.
- When the data is not normally distributed.
- When the data is ordinal or ranked.

### When Not to Use
- For paired or dependent data.
- For more than two groups.
- When the observations between groups are not independent.

### Business Example
A company wants to compare customer spending between customers from two different stores.

### Python Example
We can use `scipy.stats.mannwhitneyu()` to perform the test.

### Interpretation
- If p-value < 0.05 → Reject the Null Hypothesis.
- If p-value >= 0.05 → Fail to Reject the Null Hypothesis.
- A significant result indicates that the distributions of the two groups are statistically different.

```python
from scipy.stats import mannwhitneyu

store_a = [120, 150, 110, 180, 140, 130, 160]
store_b = [200, 220, 190, 210, 230, 205, 215]

u_statistic, p_value = mannwhitneyu(
    store_a,
    store_b,
    alternative="two-sided"
)

print("Mann-Whitney U Statistic:", u_statistic)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("The two groups are statistically different.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence of a difference between the two groups.")
```

## Business Example and Interpretation

### Business Example
A company compares customer spending from Store A and Store B.

The Mann-Whitney U Test can be used when the spending data is not normally distributed.

### Interpretation
If the p-value is less than 0.05, we reject the Null Hypothesis and conclude that there is statistical evidence that the two groups differ.

If the p-value is greater than or equal to 0.05, there is not enough evidence to conclude that the two groups differ.

### AI/ML Use Case
The Mann-Whitney U Test can be used to compare the error or performance distributions of two independent machine learning models when the assumptions of a t-test are not suitable.

## 6. Shapiro-Wilk Test

### Definition
The Shapiro-Wilk Test is a statistical test used to check whether a dataset is approximately **normally distributed**.

The Null Hypothesis (H₀) says that the data comes from a normal distribution.

### Assumptions
- The data should be numerical.
- Observations should be independent.
- The test is mainly used for continuous data.
- It works best with a reasonable sample size.

### When to Use
- To check normality before using tests like t-test or ANOVA.
- When we need to check whether a numerical dataset follows a normal distribution.

### When Not to Use
- For categorical data.
- When the data is not independent.
- For very large datasets, because the test can detect very small deviations from normality.

### Business Example
A company wants to check whether customer delivery times are approximately normally distributed before applying a statistical test.

### Python Example
We can use `scipy.stats.shapiro()` to perform the Shapiro-Wilk Test.

### Interpretation
- If p-value < 0.05 → Reject the Null Hypothesis → data is likely not normally distributed.
- If p-value >= 0.05 → Fail to Reject the Null Hypothesis → there is not enough evidence to say the data is not normally distributed.

A p-value greater than 0.05 does **not prove** that the data is perfectly normal.

```python
from scipy import stats

data = [10, 12, 11, 13, 12, 14, 11, 13, 12, 10]

statistic, p_value = stats.shapiro(data)

print("Shapiro-Wilk Statistic:", statistic)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("The data is likely not normally distributed.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence to say the data is not normally distributed.")
```

## Business Example and Interpretation

### Business Example
A delivery company checks whether delivery times follow a normal distribution before using a parametric statistical test.

### Interpretation
If the p-value is less than 0.05, we reject the Null Hypothesis and conclude that there is evidence that the data is not normally distributed.

If the p-value is greater than or equal to 0.05, we fail to reject the Null Hypothesis. This means there is not enough evidence to say that the data is not normally distributed.

### AI/ML Use Case
The Shapiro-Wilk Test can help check the normality assumption of features or model residuals before applying statistical analysis.
