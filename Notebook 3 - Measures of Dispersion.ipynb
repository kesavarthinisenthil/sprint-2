## 1. Range

### Understand the Concept

Range is a simple measure of dispersion that shows the difference between the highest and lowest values in a dataset.

It tells us how widely the values are spread.

### Formula

Range = Maximum Value - Minimum Value

### Demonstrate the Concept

Suppose the marks of five students are:

60, 70, 80, 90, 100

Maximum value = 100

Minimum value = 60

Range = 100 - 60 = 40

So, the range of the marks is 40.

```python
import numpy as np

marks = [60, 70, 80, 90, 100]

range_value = np.max(marks) - np.min(marks)

print("Maximum value:", np.max(marks))
print("Minimum value:", np.min(marks))
print("Range:", range_value)
```

### Business Example

A company can use range to understand the difference between the highest and lowest monthly sales.

For example, if monthly sales range from ₹50,000 to ₹90,000, the range is ₹40,000.

### Interpretation

The range is 40, which means the difference between the highest and lowest marks is 40.

A larger range generally indicates that the values are more spread out.

### AI/ML Use Case

Range can be used during Exploratory Data Analysis to get a quick understanding of the spread of a numerical feature.

For example, it can show the difference between the minimum and maximum values of salary, age, or house price.

## 2. Variance

### Understand the Concept

Variance measures how much the values in a dataset differ from the mean.

It is calculated by finding the average of the squared differences between each value and the mean.

A small variance means the values are closer to the mean.

A large variance means the values are more spread out.

### Formula

Population Variance = Σ(x - μ)² / N

Where:

- x = each value
- μ = population mean
- N = number of values

### Demonstrate the Concept

Suppose the data is:

2, 4, 6

Mean = (2 + 4 + 6) / 3 = 4

Squared differences:

(2 - 4)² = 4

(4 - 4)² = 0

(6 - 4)² = 4

Variance = (4 + 0 + 4) / 3 = 2.67

```python
import numpy as np

data = [2, 4, 6]

variance = np.var(data)

print("Mean:", np.mean(data))
print("Variance:", variance)
```

### Business Example

A company can use variance to understand how much its daily sales vary from the average sales.

For example, low variance in daily sales means sales are relatively consistent, while high variance means sales change more significantly.

### Interpretation

The variance is approximately 2.67.

It shows the amount of spread in the data around the mean.

Because the differences are squared, variance is expressed in squared units.

### AI/ML Use Case

Variance is useful in Exploratory Data Analysis and feature analysis.

It helps us understand how much a numerical feature varies and is also an important concept in machine learning methods such as feature selection and Principal Component Analysis (PCA).

## 3. Standard Deviation

### Understand the Concept

Standard deviation measures how much the values in a dataset are spread around the mean.

A small standard deviation means the values are close to the mean.

A large standard deviation means the values are more spread out.

Standard deviation is the square root of variance.

### Formula

Population Standard Deviation = √[Σ(x - μ)² / N]

### Demonstrate the Concept

Suppose the data is:

2, 4, 6

Mean = 4

Variance = 2.67

Standard Deviation = √2.67 ≈ 1.63

So, the standard deviation is approximately 1.63.

```python
import numpy as np

data = [2, 4, 6]

standard_deviation = np.std(data)

print("Mean:", np.mean(data))
print("Standard deviation:", standard_deviation)
```

### Business Example

A company can use standard deviation to understand how much its daily sales vary from the average sales.

A lower standard deviation means the sales are more consistent, while a higher standard deviation means the sales fluctuate more.

### Interpretation

The standard deviation is approximately 1.63.

It tells us how far the values generally spread from the mean.

Unlike variance, standard deviation is expressed in the same units as the original data.

### AI/ML Use Case

Standard deviation is commonly used during Exploratory Data Analysis and data preprocessing.

It helps identify the spread of numerical features and is also used in standardization, where features are transformed to have a mean of 0 and standard deviation of 1.

## 4. Quartiles

### Understand the Concept

Quartiles divide an ordered dataset into four equal parts.

There are three main quartiles:

- Q1 (First Quartile) → 25% of the data is below this value.
- Q2 (Second Quartile) → 50% of the data is below this value. It is also the median.
- Q3 (Third Quartile) → 75% of the data is below this value.

### Demonstrate the Concept

Suppose the data is:

10, 20, 30, 40, 50, 60, 70, 80

Using the percentile method:

Q1 = 27.5

Q2 = 45

Q3 = 62.5

These values divide the dataset into different portions based on their position.

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 80]

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1)
print("Q2 (Median):", q2)
print("Q3:", q3)
```

### Business Example

A company can use quartiles to analyze employee salaries.

For example, Q1 can show the salary level below which approximately 25% of employees fall, while Q3 shows the level below which approximately 75% of employees fall.

### Interpretation

Q1, Q2, and Q3 divide the data into four parts.

Q2 is the median because it represents the 50th percentile.

Quartiles help us understand how the data is distributed without looking at every individual value.

### AI/ML Use Case

Quartiles are useful during Exploratory Data Analysis.

They help us understand feature distributions and are also used to identify outliers using the Interquartile Range (IQR).

## 5. Percentiles

### Understand the Concept

A percentile tells us the position of a value relative to the rest of the data.

The kth percentile is the value below which approximately k% of the data falls.

For example:

- 25th percentile → approximately 25% of the data is below this value.
- 50th percentile → approximately 50% of the data is below this value.
- 75th percentile → approximately 75% of the data is below this value.

The 50th percentile is also the median.

### Demonstrate the Concept

Suppose the marks are:

10, 20, 30, 40, 50, 60, 70, 80

The 25th percentile is 27.5.

This means approximately 25% of the data falls below 27.5.

```python
import numpy as np

marks = [10, 20, 30, 40, 50, 60, 70, 80]

p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)

print("25th percentile:", p25)
print("50th percentile:", p50)
print("75th percentile:", p75)
```

### Business Example

An e-commerce company can use percentiles to understand delivery times.

For example, if the 90th percentile of delivery time is 5 days, it means approximately 90% of deliveries take 5 days or less.

### Interpretation

The 25th, 50th, and 75th percentiles help us understand different positions within the dataset.

Percentiles are useful for comparing individual values with the rest of the data.

### AI/ML Use Case

Percentiles are used in AI/ML for understanding feature distributions, detecting unusual values, and analyzing model performance.

For example, percentiles can be used to identify unusually high transaction amounts in a fraud detection dataset.

## 6. Interquartile Range (IQR)

### Understand the Concept

Interquartile Range (IQR) measures the spread of the middle 50% of the data.

It is calculated using the difference between the third quartile (Q3) and the first quartile (Q1).

### Formula

IQR = Q3 - Q1

### Demonstrate the Concept

Suppose:

Q1 = 20

Q3 = 60

Then:

IQR = 60 - 20

IQR = 40

So, the middle 50% of the data has a spread of 40.

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 80]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

### Business Example

A company can use IQR to understand the spread of employee salaries while reducing the effect of very high or very low salaries.

For example, IQR can show the salary range of the middle 50% of employees.

### Interpretation

IQR represents the spread between Q1 and Q3.

It focuses on the middle 50% of the data and is less affected by extreme values.

IQR is also commonly used to detect outliers.

### AI/ML Use Case

IQR is useful during data preprocessing to identify and handle outliers.

A common rule is:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these bounds can be considered potential outliers.

## 7. Coefficient of Variation

### Understand the Concept

Coefficient of Variation (CV) is a measure of relative variability.

It compares the standard deviation with the mean.

CV is useful when we want to compare the variation between datasets that have different means or different units.

### Formula

CV = (Standard Deviation / Mean) × 100

### Demonstrate the Concept

Suppose a dataset has:

Mean = 50

Standard Deviation = 5

CV = (5 / 50) × 100

CV = 10%

So, the coefficient of variation is 10%.

```python
import numpy as np

data = [40, 45, 50, 55, 60]

mean = np.mean(data)
standard_deviation = np.std(data)

cv = (standard_deviation / mean) * 100

print("Mean:", mean)
print("Standard deviation:", standard_deviation)
print("Coefficient of Variation:", cv, "%")
```

### Business Example

A company can use the coefficient of variation to compare the consistency of sales between two products.

For example, if Product A has a CV of 10% and Product B has a CV of 25%, Product B has greater relative variation in its sales.

### Interpretation

The coefficient of variation expresses variability as a percentage of the mean.

A lower CV indicates less relative variation, while a higher CV indicates more relative variation.

### AI/ML Use Case

CV can be useful when comparing the variability of different numerical features, especially when their scales or average values are different.

It can help during Exploratory Data Analysis to understand which features have relatively higher variation.
