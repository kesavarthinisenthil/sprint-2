## Null Hypothesis

### Definition
The Null Hypothesis, written as H₀, is a statement that there is no significant difference, effect, or relationship in the population.

It assumes that any observed difference happened due to random chance.

### Example
A company claims that the average delivery time is 30 minutes.

- H₀: The average delivery time is 30 minutes.
- H₁: The average delivery time is not 30 minutes.

### Why it is used
The null hypothesis provides a starting point for statistical testing.

We use sample data to decide whether there is enough evidence to reject H₀.

### Where it is used
- Comparing two groups
- Testing a business claim
- Evaluating changes
- Machine learning experiments

```python
import numpy as np

delivery_times = np.array([29, 31, 30, 32, 28, 30, 31, 29, 30, 31])

sample_mean = np.mean(delivery_times)

print("Delivery Times:", delivery_times)
print("Sample Mean:", sample_mean)

print("Null Hypothesis (H₀): Average delivery time = 30 minutes")
```

### Business Example
A company may want to check whether the average delivery time is still 30 minutes after making changes to its delivery process.

### Interpretation
The null hypothesis represents the assumption that there is no significant change.

Statistical testing helps determine whether the sample provides enough evidence to reject this assumption.

### AI/ML Use Case
In AI/ML experiments, the null hypothesis can be used to test whether a new model or feature produces a statistically significant improvement compared with an existing approach.

## Alternative Hypothesis

### Definition
The Alternative Hypothesis, written as H₁ or Hₐ, is a statement that there is a significant difference, effect, or relationship in the population.

It is the opposite of the Null Hypothesis.

### Example
A company claims that the average delivery time is 30 minutes.

- H₀: The average delivery time is 30 minutes.
- H₁: The average delivery time is not 30 minutes.

### Why it is used
The alternative hypothesis represents the change or effect that we want to test.

If there is enough statistical evidence against H₀, we reject H₀ in favor of H₁.

### Where it is used
- Comparing groups
- Testing business changes
- Testing new methods
- Evaluating AI/ML models

```python
import numpy as np

delivery_times = np.array([29, 31, 30, 32, 28, 30, 31, 29, 30, 31])

sample_mean = np.mean(delivery_times)

print("Delivery Times:", delivery_times)
print("Sample Mean:", sample_mean)

print("Null Hypothesis (H₀): Average delivery time = 30 minutes")
print("Alternative Hypothesis (H₁): Average delivery time ≠ 30 minutes")
```

### Business Example
A company may want to determine whether a new delivery process changes the average delivery time.

### Interpretation
The alternative hypothesis represents the expected difference or effect.

If the statistical test provides sufficient evidence against H₀, we reject H₀ and support H₁.

### AI/ML Use Case
In AI/ML, the alternative hypothesis can be used to test whether a new model performs differently from an existing model.

## p-value

### Definition
The p-value is the probability of getting the observed result, or a more extreme result, assuming that the Null Hypothesis (H₀) is true.

It helps us decide whether the sample provides enough evidence against H₀.

### Decision Rule

If p-value ≤ significance level (α):
- Reject H₀

If p-value > significance level (α):
- Fail to reject H₀

A commonly used significance level is α = 0.05.

### Numerical Example

Suppose:

p-value = 0.03
α = 0.05

Since 0.03 < 0.05, we reject H₀.

### Where it is used
p-values are used in hypothesis tests to determine whether an observed result is statistically significant.

```python
p_value = 0.03
significance_level = 0.05

print("p-value:", p_value)
print("Significance Level:", significance_level)

if p_value <= significance_level:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example
A company tests whether a new marketing strategy changes average sales.

If the p-value is 0.03 and α = 0.05, the result is statistically significant, so H₀ is rejected.

### Interpretation
A smaller p-value provides stronger evidence against the Null Hypothesis.

A p-value does not tell us the probability that H₀ is true.

### AI/ML Use Case
p-values can be used in statistical analysis to test whether relationships or differences observed in data are statistically significant.

## Confidence Interval

### Definition
A Confidence Interval (CI) is a range of values that is used to estimate an unknown population parameter from sample data.

It gives a range of plausible values for the population parameter along with a confidence level.

### Formula

Confidence Interval = Sample Mean ± Margin of Error

For a 95% confidence interval when population standard deviation is known:

CI = x̄ ± Z × (σ / √n)

Where:
- x̄ = Sample Mean
- Z = Critical value
- σ = Standard Deviation
- n = Sample Size

### Numerical Example

Suppose:

Sample Mean = 50
Standard Deviation = 10
Sample Size = 100
Z = 1.96

Margin of Error = 1.96 × (10 / √100)
                 = 1.96

Confidence Interval = 50 ± 1.96

So the 95% confidence interval is:

48.04 to 51.96

### Where it is used
Confidence intervals are used to estimate population means, proportions, and other population parameters from sample data.

```python
import math

sample_mean = 50
std = 10
sample_size = 100
z = 1.96

margin_of_error = z * (std / math.sqrt(sample_size))

lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error

print("Sample Mean:", sample_mean)
print("Margin of Error:", margin_of_error)
print("95% Confidence Interval:", (lower_bound, upper_bound))
```

### Business Example
A company can use a confidence interval to estimate the average amount customers spend based on a sample of customers.

### Interpretation
A 95% confidence interval gives a range calculated using a method that would contain the true population parameter in about 95% of repeated samples.

A wider interval indicates more uncertainty, while a narrower interval indicates a more precise estimate.

### AI/ML Use Case
Confidence intervals can be used to understand the uncertainty around statistical estimates and model performance measurements.

## Significance Level

### Definition
The Significance Level, represented by α (alpha), is the maximum probability of rejecting the Null Hypothesis when it is actually true.

It is chosen before performing a hypothesis test.

### Common Values
- α = 0.10 → 10%
- α = 0.05 → 5%
- α = 0.01 → 1%

The most commonly used significance level is 0.05.

### Numerical Example

Suppose:

p-value = 0.03
α = 0.05

Since:

0.03 < 0.05

We reject the Null Hypothesis.

### Where it is used
The significance level is used as the decision threshold when interpreting a p-value.

```python
p_value = 0.03
alpha = 0.05

print("p-value:", p_value)
print("Significance Level (α):", alpha)

if p_value <= alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example
A company tests whether a new advertising strategy has changed average sales.

If α = 0.05 and the p-value is 0.03, the result is statistically significant and H₀ is rejected.

### Interpretation
A smaller α means we require stronger evidence before rejecting H₀.

The significance level is decided before the hypothesis test.

### AI/ML Use Case
Significance levels can be used in statistical analysis to determine whether an observed relationship or difference is statistically significant.

## Type I Error

### Definition
Type I Error occurs when we reject the Null Hypothesis (H₀) even though H₀ is actually true.

In simple words, we conclude that there is an effect or difference when there is actually no effect or difference.

Type I Error is also called a **false positive**.

### Example

Suppose:

H₀: A new medicine has no effect.

If we reject H₀ and conclude that the medicine has an effect, but actually it has no effect, this is a Type I Error.

### Relationship with Significance Level

The probability of making a Type I Error is represented by α (significance level).

For example:

α = 0.05 → 5% significance level.

### Where it is used
Type I Error is important when interpreting the results of hypothesis tests.

```python
alpha = 0.05

print("Significance Level (α):", alpha)
print("Maximum probability of Type I Error:", alpha)
```

### Business Example
A company tests whether a new marketing strategy increases sales.

If there is actually no improvement, but the test concludes that the strategy improves sales, this is a Type I Error.

### Interpretation
Type I Error means detecting an effect when there is actually no effect.

The significance level α controls the allowed probability of Type I Error.

### AI/ML Use Case
Type I Error can occur when a statistical test incorrectly identifies a feature, treatment, or model improvement as significant when there is actually no real effect.

## Type II Error

### Definition
Type II Error occurs when we fail to reject the Null Hypothesis (H₀) even though H₀ is actually false.

In simple words, we conclude that there is no significant effect when an actual effect exists.

Type II Error is also called a **false negative**.

### Example

Suppose:

H₀: A new medicine has no effect.

If the medicine actually works, but the test fails to detect its effect, this is a Type II Error.

### Symbol

The probability of Type II Error is represented by β (beta).

### Where it is used
Type II Error is important when evaluating the reliability and sensitivity of a hypothesis test.

```python
beta = 0.10

print("Probability of Type II Error (β):", beta)
print("Probability of detecting the effect:", 1 - beta)
```

### Business Example
A company tests whether a new advertising strategy improves sales.

If the strategy actually improves sales, but the test fails to detect the improvement, this is a Type II Error.

### Interpretation
Type II Error means failing to detect an effect when an actual effect exists.

The probability of Type II Error is represented by β.

### AI/ML Use Case
Type II Error can occur when a statistical test fails to identify a real difference between two models, features, or treatments.

## One-tailed Test

### Definition
A one-tailed test is a hypothesis test where the alternative hypothesis looks for an effect in only one direction.

It can test whether a value is:
- Greater than a specific value, or
- Less than a specific value.

### Example

A company wants to check whether a new process increases the average production above 100 units.

- H₀: μ ≤ 100
- H₁: μ > 100

This is a right-tailed test.

If we wanted to test whether production is less than 100:

- H₀: μ ≥ 100
- H₁: μ < 100

This is a left-tailed test.

### Where it is used
One-tailed tests are used when the research question has a specific expected direction.

```python
import math

sample_mean = 105
population_mean = 100
std = 10
sample_size = 100

z_score = (sample_mean - population_mean) / (std / math.sqrt(sample_size))

print("Z-score:", z_score)
```

### Business Example
A company wants to know whether a new production method increases the average number of products produced per day.

Since the test is only checking for an increase, a one-tailed test can be used.

### Interpretation
A one-tailed test focuses on one direction:
- Right-tailed → greater than
- Left-tailed → less than

### AI/ML Use Case
A one-tailed test can be used when an experiment has a predefined directional hypothesis, such as testing whether a new method increases model performance.

## Two-tailed Test

### Definition
A two-tailed test is a hypothesis test where the alternative hypothesis checks for a difference in both directions.

It tests whether a value is:
- Greater than a specific value, or
- Less than a specific value.

### Example

A company wants to check whether the average production is different from 100 units.

- H₀: μ = 100
- H₁: μ ≠ 100

This is a two-tailed test because we are checking both:
- μ > 100
- μ < 100

### Significance Level

If α = 0.05, the significance level is divided between the two tails:

0.025 in the left tail and 0.025 in the right tail.

### Where it is used
Two-tailed tests are used when we want to detect a difference in either direction.

```python
import math

sample_mean = 105
population_mean = 100
std = 10
sample_size = 100

z_score = (sample_mean - population_mean) / (std / math.sqrt(sample_size))

print("Z-score:", z_score)

if abs(z_score) > 1.96:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example
A company wants to check whether a new production method changes the average production level.

The change could be an increase or a decrease, so a two-tailed test is appropriate.

### Interpretation
A two-tailed test checks for differences in both directions.

For a 5% significance level, the common critical values for a standard normal test are approximately -1.96 and +1.96.

### AI/ML Use Case
Two-tailed tests can be used to check whether a new model, feature, or method produces a statistically significant difference from an existing approach.
