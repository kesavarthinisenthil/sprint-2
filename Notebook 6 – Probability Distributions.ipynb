## 1. Bernoulli Distribution

### Definition

Bernoulli distribution is a probability distribution for a single experiment that has only two possible outcomes.

The two outcomes are usually called:

- Success = 1
- Failure = 0

For example, a customer either purchases a product or does not purchase it.

### Formula

P(X = x) = p^x(1 - p)^(1 - x)

Where:

- X = Random variable
- p = Probability of success
- 1 - p = Probability of failure
- x = 0 or 1

### Numerical Example

Suppose the probability that a customer purchases a product is 0.6.

Therefore:

P(Success) = 0.6

P(Failure) = 1 - 0.6 = 0.4

So, the two possible outcomes are:

Success = 1 → Probability = 0.6

Failure = 0 → Probability = 0.4

### Where It Is Used

Bernoulli distribution is used when there is one trial with two possible outcomes.

### Demonstrate the Concept

A single yes/no, success/failure, or true/false experiment can be represented using a Bernoulli distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.6

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities)
plt.xticks([0, 1], ["Failure", "Success"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.show()
```

### Real-World Example

A company can use Bernoulli distribution to represent whether a customer purchases a product or not.

Purchase = 1

No Purchase = 0

### AI/ML Example

Bernoulli distribution can represent binary outcomes such as:

- Spam / Not Spam
- Fraud / Not Fraud
- Churn / Not Churn

### Interpretation

Bernoulli distribution is useful when an experiment has exactly two possible outcomes.

## 2. Binomial Distribution

### Definition

Binomial distribution is used to find the number of successes in a fixed number of independent trials.

Each trial has only two possible outcomes: success or failure.

### Formula

P(X = k) = C(n, k) × p^k × (1 - p)^(n-k)

Where:

- n = Number of trials
- k = Number of successes
- p = Probability of success
- 1 - p = Probability of failure

### Numerical Example

Suppose 5 customers are contacted and the probability that each customer purchases a product is 0.6.

Here:

n = 5
k = 3
p = 0.6

P(X = 3) = C(5,3) × (0.6)^3 × (0.4)^2

= 10 × 0.216 × 0.16

= 0.3456

So, the probability of exactly 3 customers purchasing the product is 0.3456 or 34.56%.

### Where It Is Used

Binomial distribution is used when:

- There is a fixed number of trials.
- Each trial has two possible outcomes.
- Trials are independent.
- The probability of success remains constant.

### Demonstrate the Concept

The number of successful outcomes in multiple independent yes/no trials can be represented using a binomial distribution.

```python
import math
import numpy as np
import matplotlib.pyplot as plt

n = 5
p = 0.6

k_values = np.arange(0, n + 1)

probabilities = [
    math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in k_values
]

print("Probability of exactly 3 successes:", probabilities[3])

plt.bar(k_values, probabilities)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.xticks(k_values)
plt.show()
```

### Real-World Example

A company contacts 5 customers and wants to know the probability that exactly 3 customers will purchase a product.

### AI/ML Example

Binomial distribution can be useful for modeling the number of successful outcomes in a fixed number of binary events, such as:

- Number of successful predictions
- Number of customers responding to an offer
- Number of positive outcomes in an experiment

### Interpretation

The graph shows the probability of getting different numbers of successes out of 5 trials.

The highest probability in this example is around 3 successes.

## 3. Poisson Distribution

### Definition

Poisson distribution is used to find the probability of a certain number of events happening in a fixed time or space interval.

It is useful when events happen independently at an average rate.

### Formula

P(X = k) = (e^(-λ) × λ^k) / k!

Where:

- X = Number of events
- k = Number of events
- λ = Average number of events in the interval
- e = Euler's number

### Numerical Example

Suppose a customer support team receives an average of 3 calls per hour.

Here:

λ = 3

What is the probability of receiving exactly 2 calls in one hour?

P(X = 2) = (e^(-3) × 3^2) / 2!

= (0.0498 × 9) / 2

≈ 0.224

So, the probability of receiving exactly 2 calls is approximately 22.4%.

### Where It Is Used

Poisson distribution is used for counting the number of events occurring in a fixed interval.

Examples include:

- Number of calls per hour
- Number of website requests per minute
- Number of customers arriving at a store

### Demonstrate the Concept

Poisson distribution models how many times an event occurs within a fixed time or space interval.

```python
import math
import numpy as np
import matplotlib.pyplot as plt

lambda_value = 3

k_values = np.arange(0, 11)

probabilities = [
    (math.exp(-lambda_value) * (lambda_value ** k)) / math.factorial(k)
    for k in k_values
]

print("Probability of exactly 2 events:", probabilities[2])

plt.bar(k_values, probabilities)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.xticks(k_values)
plt.show()
```

### Real-World Example

A customer support center receives an average of 3 calls per hour. Poisson distribution can be used to estimate the probability of receiving a specific number of calls in an hour.

### AI/ML Example

Poisson distribution can be used for count-based data, such as:

- Number of customer complaints
- Number of website requests
- Number of events occurring in a time period

### Interpretation

The graph shows the probability of different numbers of events occurring when the average number of events is 3.

The probabilities are highest around the average value.

## 4. Uniform Distribution

### Definition

Uniform distribution is a probability distribution where all values within a given range have equal probability.

For a continuous uniform distribution, the values are between a minimum value `a` and a maximum value `b`.

### Formula

f(x) = 1 / (b - a)

Where:

- a = Minimum value
- b = Maximum value
- x = Value between a and b

The probability of a value falling between two points is:

P(c ≤ X ≤ d) = (d - c) / (b - a)

### Numerical Example

Suppose a value is uniformly distributed between 10 and 20.

Here:

a = 10
b = 20

The probability of the value being between 12 and 15 is:

P(12 ≤ X ≤ 15)

= (15 - 12) / (20 - 10)

= 3 / 10

= 0.3

So, the probability is 30%.

### Where It Is Used

Uniform distribution is used when every value within a specific range is equally likely.

### Demonstrate the Concept

For example, if a random value is selected between 10 and 20, values within this range have the same probability density.

```python
import numpy as np
import matplotlib.pyplot as plt

a = 10
b = 20

data = np.random.uniform(a, b, 1000)

plt.hist(data, bins=20, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Uniform Distribution")
plt.show()
```

### Real-World Example

Suppose a customer arrives at a store at any random time between 10:00 AM and 11:00 AM, and every time within this interval is considered equally likely.

This can be represented using a uniform distribution.

### AI/ML Example

Uniform distribution can be used in AI/ML for:

- Generating random data
- Random initialization
- Simulation and testing
- Random sampling

### Interpretation

The histogram shows that the values are spread approximately evenly between 10 and 20.

There is no particular value that is more likely than another within the range.

## 5. Normal Distribution

### Definition

Normal distribution is a continuous probability distribution with a bell-shaped and symmetric curve.

In a normal distribution, the mean, median, and mode are equal.

### Formula

f(x) = 1 / (σ√(2π)) × e^(-(x - μ)² / (2σ²))

Where:

- μ = Mean
- σ = Standard deviation
- x = Value
- e = Euler's number

### Numerical Example

Suppose the heights of a group of people follow a normal distribution with:

Mean (μ) = 50
Standard deviation (σ) = 10

At the mean value x = 50:

f(50) = 1 / (10√(2π))

≈ 0.0399

So, the probability density at the mean is approximately 0.0399.

### Where It Is Used

Normal distribution is commonly used when data is continuous and follows a bell-shaped pattern.

Examples include:

- Heights
- Test scores
- Measurement errors
- Many natural measurements

### Demonstrate the Concept

Most values are concentrated around the mean, while fewer values occur farther away from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std = 10

x = np.linspace(mean - 4 * std, mean + 4 * std, 1000)

pdf = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -((x - mean) ** 2) / (2 * std ** 2)
)

plt.plot(x, pdf)
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")
plt.grid()
plt.show()
```

### Real-World Example

A company's employee performance scores may follow a normal distribution, with most employees having scores close to the average.

### AI/ML Example

Normal distribution is commonly used in AI/ML for:

- Understanding feature distributions
- Data preprocessing and standardization
- Gaussian Naive Bayes
- Statistical modeling

### Interpretation

The graph has a bell-shaped and symmetric curve.

Most values are close to the mean, and the probability density decreases as we move farther away from the mean.

## 6. Exponential Distribution

### Definition

Exponential distribution is a continuous probability distribution used to model the waiting time between events.

It is commonly used when events occur independently at a constant average rate.

### Formula

f(x) = λe^(-λx)

Where:

- λ = Rate of events
- x = Waiting time
- e = Euler's number
- x ≥ 0

The probability of an event occurring within a time x is:

P(X ≤ x) = 1 - e^(-λx)

### Numerical Example

Suppose a customer support center receives an average of 0.5 calls per minute.

So:

λ = 0.5

What is the probability that the next call arrives within 2 minutes?

P(X ≤ 2) = 1 - e^(-0.5 × 2)

= 1 - e^(-1)

≈ 1 - 0.3679

≈ 0.6321

So, the probability is approximately 63.21%.

### Where It Is Used

Exponential distribution is used to model waiting time between events.

Examples include:

- Time between customer arrivals
- Time between phone calls
- Time between website requests
- Waiting time for a service

### Demonstrate the Concept

Exponential distribution helps us understand how long we may need to wait before the next event occurs.

```python
import numpy as np
import matplotlib.pyplot as plt

lambda_value = 0.5

# Generate random waiting times
data = np.random.exponential(scale=1 / lambda_value, size=1000)

plt.hist(data, bins=30, density=True)
plt.xlabel("Waiting Time")
plt.ylabel("Density")
plt.title("Exponential Distribution")
plt.show()
```

### Real-World Example

A customer support center can use exponential distribution to model the waiting time between incoming customer calls.

### AI/ML Example

Exponential distribution can be used in AI/ML for:

- Time-to-event modeling
- Survival analysis
- Modeling waiting times
- Analyzing event arrival times

### Interpretation

The graph shows that shorter waiting times are more common, while longer waiting times become less likely.

The distribution starts with a high probability density and decreases as the waiting time increases.
