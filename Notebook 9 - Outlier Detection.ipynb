## What are Outliers?

### Definition
Outliers are data points that are very different from the other values in a dataset.

They can be unusually high or unusually low compared to the rest of the data.

### Example
Consider the values:

10, 12, 11, 13, 12, 50

Here, 50 is an outlier because it is much higher than the other values.

### Why Outliers are Important
Outliers can:
- Affect the mean and standard deviation.
- Affect statistical analysis.
- Reduce the performance of some machine learning models.

### Where it is used
Outlier detection is commonly used during data preprocessing and Exploratory Data Analysis (EDA).

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([10, 12, 11, 13, 12, 50])

mean = np.mean(data)

print("Data:", data)
print("Mean:", mean)

plt.scatter(range(len(data)), data)
plt.xlabel("Data Point")
plt.ylabel("Value")
plt.title("Example of an Outlier")
plt.grid()
plt.show()
```

### Business Example
In a company's sales data, if most daily sales are between ₹10,000 and ₹20,000 but one day has ₹1,00,000, that value may be an outlier.

### Interpretation
An outlier does not always mean the data is wrong. It may represent a genuine unusual event or an error in data collection.

### AI/ML Use Case
In AI/ML, outliers are checked during data preprocessing because extreme values can affect model training and predictions.

## Types of Outliers

### Definition
Outliers can be classified based on how and where they occur in the data.

### Main Types

1. **Point Outlier**
   - A single data point is very different from the other values.
   - Example: [10, 12, 11, 13, 12, 50]
   - Here, 50 is a point outlier.

2. **Contextual Outlier**
   - A value is considered an outlier only in a particular context.
   - Example: A temperature of 35°C may be normal in summer but unusual in winter.

3. **Collective Outlier**
   - A group of data points behaves differently from the normal pattern.
   - Example: A sudden group of unusual network requests may indicate abnormal activity.

### Why it is important
Identifying the type of outlier helps us understand whether it is an error, a special event, or a meaningful pattern.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([10, 12, 11, 13, 12, 50])

plt.scatter(range(len(data)), data)

plt.xlabel("Data Point")
plt.ylabel("Value")
plt.title("Point Outlier Example")
plt.grid()
plt.show()
```

### Business Example
In sales data, one unusually high transaction can be a point outlier.

A sudden increase in sales only during a particular season can be a contextual outlier.

A continuous group of unusual transactions may be a collective outlier.

### Interpretation
The type of outlier depends on whether the unusual behavior comes from:
- One data point
- The surrounding context
- A group of data points

### AI/ML Use Case
Outlier types are useful in fraud detection, network monitoring, sensor data analysis, and anomaly detection.

## Detecting Outliers

### Definition
Outlier detection means finding data points that are very different from the normal pattern of the dataset.

Common methods used to detect outliers are:
- IQR Method
- Z-Score Method
- Box Plot
- Scatter Plot

### Example
Consider:

[10, 12, 11, 13, 12, 50]

The value 50 is much higher than the other values, so it may be an outlier.

### Why it is used
Outlier detection helps us identify unusual values before performing statistical analysis or training a machine learning model.

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 50])

mean = np.mean(data)
std = np.std(data)

print("Data:", data)
print("Mean:", mean)
print("Standard Deviation:", std)

# Simple comparison with mean
for value in data:
    if abs(value - mean) > 2 * std:
        print("Possible outlier:", value)
```

### Business Example
A company can detect unusually high or low transaction amounts in its sales data.

### Interpretation
A value that is very far from the normal pattern may be an outlier. However, we should use a proper method such as IQR or Z-score to confirm it.

### AI/ML Use Case
Outlier detection is used during data preprocessing to identify unusual observations before training a machine learning model.

## IQR Method

### Definition
The IQR (Interquartile Range) method detects outliers using the middle 50% of the data.

### Formula

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound are considered outliers.

### Numerical Example

Data: [10, 12, 13, 14, 15, 16, 18, 50]

Q1 = 12.75

Q3 = 17.50

IQR = 17.50 - 12.75 = 4.75

Lower Bound = 12.75 - (1.5 × 4.75) = 5.625

Upper Bound = 17.50 + (1.5 × 4.75) = 24.625

Since 50 is greater than 24.625, 50 is an outlier.

### Where it is used
The IQR method is commonly used for numerical data and is useful when the data is skewed or contains extreme values.

```python
import numpy as np

data = np.array([10, 12, 13, 14, 15, 16, 18, 50])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

### Business Example
A company can use the IQR method to detect unusually high transaction amounts in its sales data.

### Interpretation
Values outside the lower and upper bounds are treated as possible outliers.

### AI/ML Use Case
The IQR method can be used during data preprocessing to identify extreme values before training a machine learning model.

## Z-Score Method

### Definition
Z-score tells us how far a data point is from the mean in terms of standard deviations.

### Formula

Z = (X - μ) / σ

Where:
- X = Data point
- μ = Mean
- σ = Standard deviation

A common rule is:
- Z-score between -3 and +3 → usually considered normal
- Z-score less than -3 or greater than +3 → possible outlier

### Numerical Example

Data point X = 100

Mean μ = 50

Standard deviation σ = 10

Z = (100 - 50) / 10
Z = 5

Since 5 is greater than 3, 100 can be considered an outlier.

### Where it is used
The Z-score method is commonly used when the data is approximately normally distributed.

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 14, 50])

mean = np.mean(data)
std = np.std(data)

z_scores = (data - mean) / std

print("Data:", data)
print("Z-Scores:", z_scores)

outliers = data[np.abs(z_scores) > 3]

print("Possible Outliers:", outliers)
```

### Business Example
A company can use Z-scores to identify unusually high transaction amounts compared with the average transaction.

### Interpretation
A large positive Z-score means the value is far above the mean.

A large negative Z-score means the value is far below the mean.

Values beyond ±3 are commonly treated as possible outliers.

### AI/ML Use Case
Z-score based outlier detection can be used during data preprocessing to identify unusual observations in numerical features.

## Box Plot

### Definition
A box plot is a graphical method used to show the distribution of numerical data and identify possible outliers.

It shows:
- Minimum value
- Q1 (First Quartile)
- Median
- Q3 (Third Quartile)
- Maximum value

Outliers are usually shown as individual points outside the whiskers.

### Example

Data: [10, 12, 13, 14, 15, 16, 18, 50]

The value 50 is far away from the other values, so it may appear as an outlier in the box plot.

### Where it is used
Box plots are commonly used during Exploratory Data Analysis (EDA) to quickly understand data distribution and detect outliers.

```python
import matplotlib.pyplot as plt

data = [10, 12, 13, 14, 15, 16, 18, 50]

plt.boxplot(data)

plt.ylabel("Value")
plt.title("Box Plot for Outlier Detection")
plt.grid()
plt.show()
```

### Business Example
A company can use a box plot to identify unusually high or low sales values.

### Interpretation
The box represents the middle 50% of the data.

Points outside the whiskers are possible outliers.

### AI/ML Use Case
Box plots are useful during EDA to visually detect outliers in numerical features before applying machine learning algorithms.

## Scatter Plot

### Definition
A scatter plot is a graph that shows the relationship between two numerical variables using individual data points.

It can also help identify unusual data points that are far away from the main pattern.

### Example

Suppose we have:

X = [1, 2, 3, 4, 5, 6]
Y = [10, 12, 14, 16, 18, 50]

Most points follow an increasing pattern, but the point (6, 50) is far from the main pattern and may be an outlier.

### Where it is used
- To understand relationships between two variables.
- To identify unusual observations.
- To explore patterns during EDA.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [10, 12, 14, 16, 18, 50]

plt.scatter(x, y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Scatter Plot for Outlier Detection")
plt.grid()
plt.show()
```

### Business Example
A company can plot advertising spending against sales. If one data point is very far from the general pattern, it may need further investigation.

### Interpretation
Points close to the main pattern represent normal observations.

A point far away from the pattern may be an outlier.

### AI/ML Use Case
Scatter plots are useful during Exploratory Data Analysis (EDA) to understand relationships between features and identify unusual observations before model training.
