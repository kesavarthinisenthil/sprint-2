## 1. Sampling

### Definition

Sampling is the process of selecting a smaller group of people or data points from a larger population.

The selected group is called a sample.

Sampling is used because studying the entire population can take more time, cost more, and require more resources.

### Example

Suppose a company has 10,000 customers.

Instead of asking all 10,000 customers for feedback, the company selects 500 customers and collects their feedback.

Here:

- Population = 10,000 customers
- Sample = 500 customers

### Where It Is Used

Sampling is used when:

- The population is very large.
- Collecting data from everyone is difficult.
- We need to save time and resources.
- We want to estimate population characteristics using a sample.

### Demonstrate the Concept

A sample should represent the population as closely as possible.

A good sample helps us make useful conclusions about the larger population.

```python
import numpy as np

# Population of 1000 customers
population = np.arange(1, 1001)

# Select a sample of 100 customers
sample = np.random.choice(population, size=100, replace=False)

print("Population size:", len(population))
print("Sample size:", len(sample))
print("First 10 sampled customers:", sample[:10])
```

### Real-World Example

A company wants to understand customer satisfaction among 50,000 customers.

Instead of contacting all customers, it selects 1,000 customers and collects their feedback.

### AI/ML Example

In AI/ML, sampling can be used to select a smaller portion of a large dataset for:

- Data analysis
- Model development
- Testing
- Exploratory analysis

### Interpretation

Sampling allows us to work with a smaller amount of data while trying to understand the larger population.

The quality of the sample is important because a poor sample may not represent the population correctly.

## 2. Random Sampling

### Definition

Random sampling is a sampling method where every member of the population has an equal chance of being selected.

The selection is done randomly without intentionally choosing specific members.

### Example

Suppose a company has 1,000 customers and wants to select 100 customers for a survey.

If the 100 customers are selected randomly, every customer has an equal chance of being selected.

### Where It Is Used

Random sampling is used when:

- We want to reduce selection bias.
- The population is reasonably similar.
- Every member should have an equal chance of selection.
- We need a simple and fair sampling method.

### Demonstrate the Concept

A random sample can be selected using a random selection method.

For example, we can randomly select 100 customer IDs from a population of 1,000 customers.

```python
import numpy as np

# Population of 1000 customers
population = np.arange(1, 1001)

# Randomly select 100 customers
random_sample = np.random.choice(
    population,
    size=100,
    replace=False
)

print("Population size:", len(population))
print("Sample size:", len(random_sample))
print("First 10 selected customers:", random_sample[:10])
```

### Real-World Example

A company wants feedback from its customers.

Instead of selecting only customers from one location or one age group, it randomly selects customers from the entire customer list.

### AI/ML Example

Random sampling can be used to select data points from a large dataset for:

- Model training
- Testing
- Data analysis
- Creating representative subsets

### Interpretation

Random sampling gives each population member an equal chance of being selected.

This helps reduce the chance of selecting a biased sample.

## 3. Stratified Sampling

### Definition

Stratified sampling is a sampling method where the population is divided into smaller groups called strata.

Then, samples are selected from each group.

The groups are usually created based on characteristics such as:

- Age
- Gender
- Location
- Education
- Income

### Example

Suppose a company has 1,000 customers:

- 600 are from urban areas
- 400 are from rural areas

If we want a sample of 100 customers, we can select:

- 60 urban customers
- 40 rural customers

This keeps the same proportion as the population.

### Where It Is Used

Stratified sampling is useful when the population contains different groups and we want each important group to be represented in the sample.

### Demonstrate the Concept

The population is divided into groups first.

Then, a sample is randomly selected from each group.

```python
import pandas as pd

# Create a population
data = pd.DataFrame({
    "Customer_ID": range(1, 11),
    "Group": [
        "Urban", "Urban", "Urban", "Urban", "Urban",
        "Rural", "Rural", "Rural", "Rural", "Rural"
    ]
})

# Select samples from each group
stratified_sample = (
    data.groupby("Group", group_keys=False)
    .apply(lambda x: x.sample(2, random_state=42))
)

print("Population:")
print(data)

print("\nStratified Sample:")
print(stratified_sample)
```

### Real-World Example

A company wants to survey customers from different regions.

It divides customers into groups such as North, South, East, and West and selects customers from each region.

### AI/ML Example

Stratified sampling can be used when a dataset contains important groups that need to be represented in training or testing data.

For example, a dataset may contain customers from different regions, and we want each region to be represented.

### Interpretation

Stratified sampling ensures that important groups in the population are included in the sample.

It is useful when different groups may have different characteristics.

## 4. Cluster Sampling

### Definition

Cluster sampling is a sampling method where the population is divided into groups called clusters.

Instead of selecting individuals from every cluster, some clusters are randomly selected, and the members of the selected clusters are studied.

### Example

Suppose a company has customers from 20 cities.

Each city can be considered a cluster.

Instead of selecting customers from all 20 cities, the company randomly selects 5 cities and collects data from customers in those selected cities.

### Where It Is Used

Cluster sampling is useful when:

- The population is spread across different locations.
- It is difficult or expensive to reach individual members.
- Groups or clusters already exist.
- Data collection needs to be simpler and more cost-effective.

### Demonstrate the Concept

The population is first divided into clusters.

Then, some clusters are randomly selected for data collection.

```python
import numpy as np

# Population divided into 5 clusters
clusters = {
    "City_A": list(range(1, 11)),
    "City_B": list(range(11, 21)),
    "City_C": list(range(21, 31)),
    "City_D": list(range(31, 41)),
    "City_E": list(range(41, 51))
}

# Randomly select 2 clusters
np.random.seed(42)
selected_clusters = np.random.choice(
    list(clusters.keys()),
    size=2,
    replace=False
)

# Collect all members from selected clusters
cluster_sample = []

for cluster in selected_clusters:
    cluster_sample.extend(clusters[cluster])

print("Selected Clusters:", selected_clusters)
print("Cluster Sample:", cluster_sample)
```

### Real-World Example

A company wants to study customer satisfaction across many cities.

Instead of visiting every city, it randomly selects a few cities and surveys customers in those selected cities.

### AI/ML Example

Cluster sampling can be useful when a large dataset is naturally divided into groups such as:

- Cities
- Stores
- Branches
- Schools
- Hospitals

A few groups can be selected to create a smaller dataset for analysis.

### Interpretation

Cluster sampling selects entire groups instead of selecting individual members from every group.

It can reduce the time and cost required to collect data when the population is geographically or naturally grouped.

## 5. Systematic Sampling

### Definition

Systematic sampling is a sampling method where we select members from a population at a fixed interval.

First, we choose a starting point randomly. Then, we select every kth member from the population.

### Formula

Sampling interval:

k = N / n

Where:

- N = Population size
- n = Required sample size
- k = Sampling interval

### Numerical Example

Suppose a company has 1,000 customers and wants to select 100 customers.

k = 1000 / 100

k = 10

If the random starting point is customer 4, we select:

4, 14, 24, 34, 44, ...

### Where It Is Used

Systematic sampling is useful when:

- The population has an ordered list.
- We need a simple sampling method.
- We want to select samples at regular intervals.
- The population is large.

### Demonstrate the Concept

We select a random starting point and then choose every kth member until the required sample size is reached.

```python
import numpy as np

# Population
population = np.arange(1, 101)

# Required sample size
sample_size = 10

# Calculate sampling interval
k = len(population) // sample_size

# Random starting point
np.random.seed(42)
start = np.random.randint(0, k)

# Select every kth member
systematic_sample = population[start::k][:sample_size]

print("Sampling interval:", k)
print("Starting position:", start + 1)
print("Systematic Sample:", systematic_sample)
```

### Real-World Example

A company has a list of 10,000 customers and wants to survey 1,000 customers.

It can select every 10th customer after choosing a random starting point.

### AI/ML Example

Systematic sampling can be used to select records from large datasets at regular intervals.

For example, every 10th transaction can be selected for analysis.

### Interpretation

Systematic sampling selects data points using a fixed interval.

It is simple and efficient, especially when the population is stored in an ordered list.

## 6. Sampling Bias

### Definition

Sampling bias occurs when the selected sample does not properly represent the population.

This can happen when some groups have a higher or lower chance of being selected.

As a result, conclusions from the sample may not accurately represent the whole population.

### Example

Suppose a company wants to know how satisfied all customers are.

If the company collects feedback only from its most frequent customers, the sample may not represent all customers.

### Where It Is Used

Understanding sampling bias is important when:

- Collecting survey data
- Building datasets
- Conducting research
- Training AI/ML models

### Demonstrate the Concept

A sample can be biased if the selection process systematically leaves out certain groups.

For example, selecting only urban customers when the population contains both urban and rural customers can create sampling bias.

```python
import numpy as np

# Population: 70 urban and 30 rural customers
population = np.array(
    ["Urban"] * 70 + ["Rural"] * 30
)

# Biased sample: selecting only from the first 70 customers
biased_sample = population[:20]

print("Population distribution:")
print("Urban:", np.sum(population == "Urban"))
print("Rural:", np.sum(population == "Rural"))

print("\nBiased sample distribution:")
print("Urban:", np.sum(biased_sample == "Urban"))
print("Rural:", np.sum(biased_sample == "Rural"))
```

### Real-World Example

A company conducts a customer survey only during working hours.

Customers who are unavailable during those hours may be left out, causing the sample to differ from the overall customer population.

### AI/ML Example

Sampling bias can affect AI/ML models when training data does not properly represent the real-world population.

For example, if a customer dataset contains mostly one customer group, the model may perform differently for underrepresented groups.

### Interpretation

Sampling bias can lead to misleading conclusions because the sample does not accurately represent the population.

A carefully designed sampling method helps reduce this problem.

## 7. Sampling Error

### Definition

Sampling error is the difference between a value calculated from a sample and the actual value of the entire population.

It happens because we study only a sample instead of the complete population.

### Formula

Sampling Error = Sample Statistic - Population Parameter

For example:

Sampling Error = Sample Mean - Population Mean

### Numerical Example

Suppose the actual average income of a population is ₹30,000.

A sample of people gives an average income of ₹28,000.

Sampling Error:

= ₹28,000 - ₹30,000

= -₹2,000

The absolute sampling error is ₹2,000.

### Where It Is Used

Sampling error is considered when:

- Estimating population values
- Conducting surveys
- Performing statistical analysis
- Building models using samples

### Demonstrate the Concept

Different samples taken from the same population can produce slightly different results.

The difference between the sample result and the actual population value is sampling error.

```python
import numpy as np

# Population data
population = np.array([10, 20, 30, 40, 50, 60, 70, 80])

# Population mean
population_mean = np.mean(population)

# Select a sample
sample = np.array([10, 20, 40, 50])

# Sample mean
sample_mean = np.mean(sample)

# Sampling error
sampling_error = sample_mean - population_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
print("Absolute Sampling Error:", abs(sampling_error))
```

### Real-World Example

A company wants to estimate the average customer spending.

The actual population average may be ₹2,500, while a sample of customers may give an average of ₹2,400.

The difference is the sampling error.

### AI/ML Example

Sampling error can occur when a model is developed using a sample of a much larger dataset.

The sample statistics may differ slightly from the statistics of the full dataset.

### Interpretation

Sampling error does not necessarily mean that the sampling process was biased.

It occurs naturally because only part of the population is observed.

Increasing the sample size generally helps reduce sampling error.

## 8. Sample Size

### Definition

Sample size is the number of observations or data points selected from a population for a study or analysis.

Choosing an appropriate sample size is important because a very small sample may not represent the population well, while a very large sample may require more time and resources.

### Formula

A commonly used formula for estimating sample size is:

n = (Z² × p × (1 - p)) / E²

Where:

- n = Required sample size
- Z = Z-score for the chosen confidence level
- p = Estimated proportion of the population
- E = Margin of error

### Numerical Example

Suppose:

- Confidence level = 95%
- Z = 1.96
- p = 0.5
- Margin of error = 0.05

Then:

n = (1.96² × 0.5 × 0.5) / 0.05²

n ≈ 384.16

So, the required sample size is approximately:

n = 385

### Where It Is Used

Sample size is important in:

- Surveys
- Statistical studies
- Experiments
- Data analysis
- AI/ML dataset preparation

### Demonstrate the Concept

Sample size depends on factors such as confidence level, margin of error, population variability, and the purpose of the study.

A suitable sample size helps us obtain more reliable results without unnecessarily collecting excessive data.

```python
import math

# Given values
z = 1.96
p = 0.5
e = 0.05

# Calculate sample size
sample_size = (z ** 2 * p * (1 - p)) / (e ** 2)

# Round up to the next whole number
required_sample_size = math.ceil(sample_size)

print("Calculated sample size:", sample_size)
print("Required sample size:", required_sample_size)
```

### Real-World Example

A company wants to conduct a customer satisfaction survey.

Instead of asking every customer, the company calculates an appropriate sample size and collects responses from that number of customers.

### AI/ML Example

Sample size is important when collecting training or testing data.

A sufficiently large and representative dataset can help an ML model learn patterns more reliably.

### Interpretation

Choosing the right sample size helps balance accuracy and resources.

A sample that is too small may give less reliable estimates, while a very large sample may require unnecessary time and cost.
